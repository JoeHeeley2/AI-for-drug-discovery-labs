 Coursework Task
----

The aim of this task is to investigate the use of machine learning to predict the inhibitory performance of compounds relative to the BRD4 protein, an important biological target for cancer treatment. The prediction target for this task is the $pIC50$, the negative base 10 logarithm of the half maximal inhibitory concentration ($IC50$). This value measures the potency of a protein inhibitor, with larger numerical values denoting more potent inhibitors. #

Using the provided dataset *Initial_PP_Data_excl.csv*, you should identify the best performing model and detail your findings and methodology into a 500-word report (no longer than 5 pages including figures and tables). You should consider:

* Dataset splits
* Different models
* Performance evaluation
* Model Comparison
 

Please read and adhere to the “Guidance on AI use for laboratory workshop reports in CHEM4032” on the module moodle page and `Acceptable_Use_of_AI_in_CHEM4032_workshops.pdf` in this directory.
