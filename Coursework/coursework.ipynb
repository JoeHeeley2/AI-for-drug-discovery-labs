# Coursework Task
----

The aim of this task is to investigate the use of machine learning to predict the inhibitory performance of compounds relative to the BRD4 protein, an important biological target for cancer treatment. The prediction target for this task is the $pIC50$, the negative base 10 logarithm of the half maximal inhibitory concentration ($IC50$). This value measure the potency of a protein inhibitor, with lower values denoting more potent inhibitors. #

Using the provided dataset *Initial_PP_Data_excl.csv*, you must find the best performing model and detail your finding and methodology into a xxx word report. You must consider:

* Dataset splits
* Different models
* Performance evaluation
* Model Comparison
